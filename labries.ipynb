## random module(np.random)

```python
# Generates 10 random numbers from a standard normal distribution (mean = 0, standard deviation = 1)
np.random.randn(10)
```

```python
# Generates 10 random integers between 10 (inclusive) and 50 (exclusive)
np.random.randint(10,50,10)
```

```python
# Generates 10 random decimal numbers between 0 and 5
np.random.uniform(0,5,10)
```
